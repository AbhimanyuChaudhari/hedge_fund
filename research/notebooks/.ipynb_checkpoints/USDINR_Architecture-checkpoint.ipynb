```python
import sys
sys.path.insert(0, r'C:\projects\hedge_fund')

import numpy as np
import pandas as pd
from data.store.duckdb_client import DuckDBClient
from data.store.s3_client import S3Client

# ── Load data ──────────────────────────────────────────────────────
db = DuckDBClient()
s  = S3Client()

files = s.list_files('processed/1s/USDINR26SEPFUT/')
dates = sorted([f.split('/')[-1].replace('.parquet','') for f in files])

all_df = []
for d in dates:
    df = db.read_parquet(f'processed/1s/USDINR26SEPFUT/{d}.parquet')
    df['ist_sec'] = (df['ts_sec'] + 19800) % 86400
    df = df[(df['ist_sec'] >= 9*3600) & (df['ist_sec'] <= 17*3600)].reset_index(drop=True)
    df['date'] = d
    all_df.append(df)
db.close()
df = pd.concat(all_df, ignore_index=True).reset_index(drop=True)
print(f'Loaded: {len(df):,} bars across {len(dates)} days')

# ── Build signals per day ──────────────────────────────────────────
TICK=0.0025; W_sig=30; NW=300

signal_dfs = []
for d in sorted(df['date'].unique()):
    sub = df[df['date']==d].copy()
    sub['bid_chg']    = sub['total_bid_qty'].diff()
    sub['ask_chg']    = sub['total_ask_qty'].diff()
    pu                = sub['close'].diff().abs() < TICK*0.5
    sub['cancel_imb'] = 0.0
    sub.loc[pu&(sub['bid_chg']<0),'cancel_imb'] -= sub['bid_chg'].abs()
    sub.loc[pu&(sub['ask_chg']<0),'cancel_imb'] += sub['ask_chg'].abs()
    sub['s_cancel']   = -sub['cancel_imb'].rolling(W_sig).mean()
    sub['s_imb']      =  sub['imbalance_last'].rolling(W_sig).mean()
    sub['mid']        = (sub['bid_p1']+sub['ask_p1'])/2
    sub['s_wmid']     = (sub['weighted_mid']-sub['mid']).rolling(10).mean()
    sub['s_momentum'] = -sub['close'].pct_change(30)
    for col in ['s_cancel','s_imb','s_wmid','s_momentum']:
        rs=sub[col].rolling(NW).std()+1e-9
        sub[col]=(sub[col]/rs).clip(-1,1)
    sub['score']     = (0.393*sub['s_cancel'] - 0.218*sub['s_imb']
                      - 0.320*sub['s_wmid']   - 0.069*sub['s_momentum'])
    sub['fwd_1800s'] = sub['close'].shift(-1800) - sub['close']
    sub['ret']       = sub['close'].diff()
    sub['spread']    = sub['ask_p1'] - sub['bid_p1']
    sub['hour']      = sub['ist_sec'] // 3600
    signal_dfs.append(sub)

df_sig = pd.concat(signal_dfs, ignore_index=True).reset_index(drop=True)
print(f'Signal df: {len(df_sig):,} bars')

# ── Regime detector ────────────────────────────────────────────────
def label_regime(df, window=1800):
    df = df.copy()
    df['price_range'] = (df['close'].rolling(window).max() -
                         df['close'].rolling(window).min())
    df['spread_z']    = ((df['spread'] - df['spread'].rolling(window).mean()) /
                         (df['spread'].rolling(window).std() + 1e-9))
    df['vol_ratio']   = (df['tick_count'] /
                         (df['tick_count'].rolling(window).mean() + 1e-9))
    conditions = [
        (df['price_range'] > 0.08) | (df['spread_z'] > 2.0),
        (df['price_range'] < 0.02) & (df['vol_ratio'] < 0.5),
    ]
    df['regime'] = np.select(conditions, ['TRENDING','DEAD'], default='MR')
    return df

result_dfs = []
for d in sorted(df_sig['date'].unique()):
    sub = df_sig[df_sig['date']==d].copy()
    sub = label_regime(sub)
    result_dfs.append(sub)

df_regime = pd.concat(result_dfs, ignore_index=True)

# ── Results ────────────────────────────────────────────────────────
print('\n=== Regime Distribution ===')
print(df_regime['regime'].value_counts())

df_val = df_regime.dropna(subset=['score','fwd_1800s'])
print('\n=== IC per Regime ===')
print(f'{"Regime":<12} {"IC_1800s":>10} {"n_bars":>8} {"pct":>6}')
print('-'*38)
for r in ['MR','TRENDING','DEAD']:
    mask = df_val['regime'] == r
    if mask.sum() < 10:
        continue
    ic  = df_val[mask]['score'].corr(df_val[mask]['fwd_1800s'])
    pct = mask.mean()*100
    print(f'{r:<12} {ic:>10.4f} {mask.sum():>8} {pct:>5.1f}%')

print('\n=== IC per Regime + Hour ===')
print(f'{"Hour":<6} {"MR_IC":>8} {"TREND_IC":>10} {"MR_n":>6}')
print('-'*34)
for h in range(9,17):
    mr_mask  = (df_val['regime']=='MR') & (df_val['hour']==h)
    trd_mask = (df_val['regime']=='TRENDING') & (df_val['hour']==h)
    ic_mr    = df_val[mr_mask]['score'].corr(df_val[mr_mask]['fwd_1800s'])
    ic_trd   = df_val[trd_mask]['score'].corr(df_val[trd_mask]['fwd_1800s']) if trd_mask.sum()>10 else np.nan
    print(f'{h}:00   {ic_mr:>8.4f} {ic_trd:>10.4f} {mr_mask.sum():>6}')
```

```python
# Trade only when ALL conditions met:
# 1. Hour is 12pm-14pm IST (best IC window)
# 2. Regime = MR (price range not too large)
# 3. Score > 0.5

print('=== Best Trading Window ===')
best = df_val[
    (df_val['hour'].isin([12, 13, 14])) &
    (df_val['regime'] == 'MR')
]
print(f'Bars in window: {len(best):,} ({len(best)/len(df_val)*100:.1f}% of day)')
print(f'IC_1800s: {best["score"].corr(best["fwd_1800s"]):.4f}')

long_signals = best[best['score'] > 0.5]
print(f'LONG signals: {len(long_signals)}')
print(f'Win rate: {(long_signals["fwd_1800s"] > 0).mean()*100:.1f}%')
print(f'Avg move: {long_signals["fwd_1800s"].mean():.5f} paise')
```

```python
# ── Realistic backtest ─────────────────────────────────────────────
# Rules:
#   Only trade 1pm-3pm IST
#   SHORT when score < -0.6
#   Hold until: score crosses 0 OR 30 min time stop
#   One trade at a time (no overlapping positions)
#   20 lots USDINR

LOT_SIZE   = 1000
LOTS       = 20
TICK       = 0.0025
THRESHOLD  = -0.6
HOLD_MAX   = 1800  # 30 min
TRADE_HOURS = [13, 14, 15]

all_trades = []

for d in sorted(df_regime['date'].unique()):
    day = df_regime[df_regime['date']==d].copy().reset_index(drop=True)
    day['hour'] = day['ist_sec'] // 3600

    inv = 0; ep = 0.0; et = 0; cash = 0.0

    for i, row in day.iterrows():
        if pd.isna(row['score']):
            continue

        ist    = int(row['ist_sec'])
        hour   = int(row['hour'])
        score  = float(row['score'])
        price  = float(row['close'])
        bid_p1 = float(row['bid_p1'])
        ask_p1 = float(row['ask_p1'])

        # Force close at 3:55pm
        if ist >= 15*3600+55*60 and inv != 0:
            xp   = ask_p1  # cover short
            pnl  = (ep - xp) * LOTS * LOT_SIZE
            cash += pnl
            all_trades.append({
                'date': d, 'hour': hour,
                'entry': ep, 'exit': xp,
                'pnl': pnl, 'reason': 'EOD',
                'hold': ist - et
            })
            inv = 0
            continue

        # Exit open SHORT
        if inv == -1:
            # Exit if signal reverses OR time stop
            if score > 0.0 or (ist - et) > HOLD_MAX:
                xp   = ask_p1
                pnl  = (ep - xp) * LOTS * LOT_SIZE
                cash += pnl
                reason = 'SIGNAL' if score > 0.0 else 'TIME'
                all_trades.append({
                    'date': d, 'hour': hour,
                    'entry': ep, 'exit': xp,
                    'pnl': pnl, 'reason': reason,
                    'hold': ist - et
                })
                inv = 0

        # Entry
        if inv == 0 and hour in TRADE_HOURS:
            if score < THRESHOLD:
                ep  = bid_p1  # short at bid
                et  = ist
                inv = -1

print(f'Total trades: {len(all_trades)}')
if all_trades:
    df_trades = pd.DataFrame(all_trades)
    pnls      = df_trades['pnl']
    wins      = pnls[pnls > 0]
    print(f'Win rate:     {len(wins)/len(pnls)*100:.1f}%')
    print(f'Total PnL:    Rs.{pnls.sum():,.0f}')
    print(f'Avg PnL:      Rs.{pnls.mean():,.0f}')
    print(f'Max win:      Rs.{pnls.max():,.0f}')
    print(f'Max loss:     Rs.{pnls.min():,.0f}')
    print(f'Avg hold:     {df_trades["hold"].mean():.0f}s')
    print()
    print('Per day:')
    for d, g in df_trades.groupby('date'):
        wr = (g['pnl']>0).mean()*100
        print(f'  {d}: {len(g):>3} trades  win={wr:.0f}%  Rs.{g["pnl"].sum():>8,.0f}')
```

```python
# Fixed backtest — time exit only
LOT_SIZE    = 1000
LOTS        = 20
TICK        = 0.0025
THRESHOLD   = -0.6
HOLD_MAX    = 1800   # 30 min hold
TRADE_HOURS = [13, 14, 15]

all_trades = []

for d in sorted(df_regime['date'].unique()):
    day  = df_regime[df_regime['date']==d].copy().reset_index(drop=True)
    day['hour'] = day['ist_sec'] // 3600

    inv = 0; ep = 0.0; et = 0; cash = 0.0

    for i, row in day.iterrows():
        if pd.isna(row['score']):
            continue

        ist    = int(row['ist_sec'])
        hour   = int(row['hour'])
        score  = float(row['score'])
        price  = float(row['close'])
        bid_p1 = float(row['bid_p1'])
        ask_p1 = float(row['ask_p1'])

        # Force close at 3:55pm
        if ist >= 15*3600+55*60 and inv != 0:
            xp   = ask_p1
            pnl  = (ep - xp) * LOTS * LOT_SIZE
            cash += pnl
            all_trades.append({
                'date': d, 'hour': hour,
                'entry': ep, 'exit': xp,
                'pnl': pnl, 'reason': 'EOD',
                'hold': ist - et
            })
            inv = 0
            continue

        # Exit on time stop only
        if inv == -1 and (ist - et) >= HOLD_MAX:
            xp   = ask_p1
            pnl  = (ep - xp) * LOTS * LOT_SIZE
            cash += pnl
            all_trades.append({
                'date': d, 'hour': hour,
                'entry': ep, 'exit': xp,
                'pnl': pnl, 'reason': 'TIME',
                'hold': ist - et
            })
            inv = 0

        # Entry — only enter once per signal cluster
        if inv == 0 and hour in TRADE_HOURS:
            if score < THRESHOLD:
                ep  = bid_p1
                et  = ist
                inv = -1

print(f'Total trades: {len(all_trades)}')
if all_trades:
    df_trades = pd.DataFrame(all_trades)
    pnls = df_trades['pnl']
    wins = pnls[pnls > 0]
    print(f'Win rate:     {len(wins)/len(pnls)*100:.1f}%')
    print(f'Total PnL:    Rs.{pnls.sum():,.0f}')
    print(f'Avg PnL:      Rs.{pnls.mean():,.0f}')
    print(f'Max win:      Rs.{pnls.max():,.0f}')
    print(f'Max loss:     Rs.{pnls.min():,.0f}')
    print(f'Avg hold:     {df_trades["hold"].mean():.0f}s')
    print()
    print('Per day:')
    for d, g in df_trades.groupby('date'):
        wr = (g['pnl']>0).mean()*100
        print(f'  {d}: {len(g):>3} trades  win={wr:.0f}%  Rs.{g["pnl"].sum():>8,.0f}')
```

```python
LOT_SIZE    = 1000
LOTS        = 20
THRESHOLD   = -0.6
HOLD_MAX    = 1800
TRADE_HOURS = [13, 14, 15]
EOD_IST     = 15*3600 + 55*60  # 3:55pm

all_trades = []

for d in sorted(df_regime['date'].unique()):
    day  = df_regime[df_regime['date']==d].copy().reset_index(drop=True)
    day['hour'] = day['ist_sec'] // 3600

    inv = 0; ep = 0.0; et = 0

    for i, row in day.iterrows():
        if pd.isna(row['score']):
            continue

        ist    = int(row['ist_sec'])
        hour   = int(row['hour'])
        score  = float(row['score'])
        bid_p1 = float(row['bid_p1'])
        ask_p1 = float(row['ask_p1'])

        # Force close at EOD — then stop trading for the day
        if ist >= EOD_IST:
            if inv == -1:
                xp  = ask_p1
                pnl = (ep - xp) * LOTS * LOT_SIZE
                all_trades.append({
                    'date': d, 'hour': hour,
                    'entry': ep, 'exit': xp,
                    'pnl': pnl, 'reason': 'EOD',
                    'hold': ist - et
                })
                inv = 0
            break  # ← stop processing this day

        # Exit on time stop
        if inv == -1 and (ist - et) >= HOLD_MAX:
            xp  = ask_p1
            pnl = (ep - xp) * LOTS * LOT_SIZE
            all_trades.append({
                'date': d, 'hour': hour,
                'entry': ep, 'exit': xp,
                'pnl': pnl, 'reason': 'TIME',
                'hold': ist - et
            })
            inv = 0

        # Entry
        if inv == 0 and hour in TRADE_HOURS and score < THRESHOLD:
            ep  = ask_p1  # short at ask (more realistic)
            et  = ist
            inv = -1

# Results
print(f'Total trades: {len(all_trades)}')
if all_trades:
    df_trades = pd.DataFrame(all_trades)
    pnls = df_trades['pnl']
    wins = pnls[pnls > 0]
    print(f'Win rate:     {len(wins)/len(pnls)*100:.1f}%')
    print(f'Total PnL:    Rs.{pnls.sum():,.0f}')
    print(f'Avg PnL:      Rs.{pnls.mean():,.0f}')
    print(f'Max win:      Rs.{pnls.max():,.0f}')
    print(f'Max loss:     Rs.{pnls.min():,.0f}')
    print(f'Avg hold:     {df_trades["hold"].mean():.0f}s')
    print()
    print('Per day:')
    for d, g in df_trades.groupby('date'):
        wr = (g['pnl']>0).mean()*100
        print(f'  {d}: {len(g):>3} trades  win={wr:.0f}%  Rs.{g["pnl"].sum():>8,.0f}')
```

```python
# Filter: only trade if PM is volatile enough
# (range expanding = mean reversion opportunities exist)

print('=== Volatility filter test ===')
print('Rule: Only trade if 1hr rolling range > 0.05 paise at entry')
print()

VOL_THRESHOLD = 0.05  # minimum range needed to trade
all_trades_filtered = []

for d in sorted(df_regime['date'].unique()):
    day  = df_regime[df_regime['date']==d].copy().reset_index(drop=True)
    day['hour']        = day['ist_sec'] // 3600
    day['roll_range']  = (day['close'].rolling(3600).max() -
                          day['close'].rolling(3600).min())

    inv = 0; ep = 0.0; et = 0

    for i, row in day.iterrows():
        if pd.isna(row['score']):
            continue

        ist        = int(row['ist_sec'])
        hour       = int(row['hour'])
        score      = float(row['score'])
        bid_p1     = float(row['bid_p1'])
        ask_p1     = float(row['ask_p1'])
        roll_range = float(row['roll_range']) if not pd.isna(row['roll_range']) else 0

        if ist >= 15*3600+55*60:
            if inv == -1:
                xp  = ask_p1
                pnl = (ep - xp) * LOTS * LOT_SIZE
                all_trades_filtered.append({
                    'date': d, 'hour': hour,
                    'entry': ep, 'exit': xp,
                    'pnl': pnl, 'reason': 'EOD',
                    'hold': ist - et
                })
                inv = 0
            break

        if inv == -1 and (ist - et) >= HOLD_MAX:
            xp  = ask_p1
            pnl = (ep - xp) * LOTS * LOT_SIZE
            all_trades_filtered.append({
                'date': d, 'hour': hour,
                'entry': ep, 'exit': xp,
                'pnl': pnl, 'reason': 'TIME',
                'hold': ist - et
            })
            inv = 0

        # Entry with volatility filter
        if inv == 0 and hour in TRADE_HOURS:
            if score < THRESHOLD and roll_range > VOL_THRESHOLD:
                ep  = ask_p1
                et  = ist
                inv = -1

print(f'Total trades: {len(all_trades_filtered)}')
if all_trades_filtered:
    df_f  = pd.DataFrame(all_trades_filtered)
    pnls  = df_f['pnl']
    wins  = pnls[pnls > 0]
    print(f'Win rate:     {len(wins)/len(pnls)*100:.1f}%')
    print(f'Total PnL:    Rs.{pnls.sum():,.0f}')
    print(f'Avg PnL:      Rs.{pnls.mean():,.0f}')
    print(f'Max win:      Rs.{pnls.max():,.0f}')
    print(f'Max loss:     Rs.{pnls.min():,.0f}')
    print()
    print('Per day:')
    for d, g in df_f.groupby('date'):
        wr = (g['pnl']>0).mean()*100
        print(f'  {d}: {len(g):>3} trades  win={wr:.0f}%  Rs.{g["pnl"].sum():>8,.0f}')
```

```python
# In Jupyter — check USDINR futures basis
# RBI publishes reference rate daily at 1:30pm IST

import requests
import pandas as pd

# Get RBI reference rate (public API)
# We'll use a simple approach — check NSE spot vs futures

# Load our futures data
db = DuckDBClient()
df_fut = db.read_parquet('processed/1s/USDINR26OCTFUT/2026-09-30.parquet')
db.close()

df_fut['ist_sec'] = (df_fut['ts_sec'] + 19800) % 86400
df_fut = df_fut[(df_fut['ist_sec'] >= 9*3600) & 
                (df_fut['ist_sec'] <= 17*3600)].reset_index(drop=True)

# OCTFUT expires Oct 28
# Days to expiry on Sep 30 = 28 days
T = 28/365

# Theoretical futures price = Spot × e^((r_INR - r_USD) × T)
# r_INR ≈ 6.5% (repo rate)
# r_USD ≈ 5.0% (fed funds rate)
# carry = 1.5% per year

r_carry = 0.015  # INR - USD rate differential

# At 1:30pm IST RBI publishes reference rate
# Let's approximate: futures at 1:30pm should be close to spot + carry
pm_data = df_fut[df_fut['ist_sec'].between(13*3600+25*60, 13*3600+35*60)]
if len(pm_data) > 0:
    fut_price = pm_data['close'].mean()
    # Theoretical spot = futures / e^(carry × T)
    implied_spot = fut_price / (1 + r_carry * T)
    basis = fut_price - implied_spot
    print(f'Futures price at 1:30pm: {fut_price:.4f}')
    print(f'Implied spot:            {implied_spot:.4f}')
    print(f'Basis (carry):           {basis:.4f} paise')
    print(f'Theoretical carry:       {implied_spot * r_carry * T:.4f} paise')
    print(f'Basis deviation:         {basis - implied_spot * r_carry * T:.4f} paise')
```

```python
spot_approx   = 95.90
monthly_carry = spot_approx * 0.015 * (30/365)
sepfut_last   = 95.7550
octfut_first  = 96.3000

actual_jump = octfut_first - sepfut_last

print(f'Theoretical 1-month carry: {monthly_carry:.4f} paise')
print(f'Actual SEPFUT→OCTFUT jump: {actual_jump:.4f} paise')
print(f'Deviation:                 {actual_jump - monthly_carry:.4f} paise')
print()
print('=== Interpretation ===')
if actual_jump > monthly_carry * 1.5:
    print('OCTFUT expensive vs carry → sell OCTFUT, buy SEPFUT')
elif actual_jump < monthly_carry * 0.5:
    print('OCTFUT cheap vs carry → buy OCTFUT, sell SEPFUT')
else:
    print('Spread within normal carry range — no arbitrage')
```
