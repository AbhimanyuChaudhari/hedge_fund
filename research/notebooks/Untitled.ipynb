```python
@njit
def e_step(logb, A, pi, starts, ends):
    # Scaled forward-backward over independent trading-day sequences.
    T,K = logb.shape
    gamma = np.empty((T,K))
    transitions = np.zeros((K,K))
    pi_counts = np.zeros(K)
    ll = 0.0
    for seq in range(len(starts)):
        lo,hi = starts[seq],ends[seq]
        n = hi-lo
        alpha = np.zeros((n,K))
        beta = np.ones((n,K))
        b = np.empty((n,K))
        scale = np.empty(n)
        for t in range(n):
            peak = np.max(logb[lo+t])
            for k in range(K): b[t,k] = np.exp(logb[lo+t,k]-peak)
            for j in range(K):
                if t == 0: alpha[t,j] = pi[j] * b[t,j]
                else:
                    v = 0.0
                    for i in range(K): v += alpha[t-1,i] * A[i,j]
                    alpha[t,j] = b[t,j] * v
            s = np.sum(alpha[t])
            scale[t] = max(s,1e-300)
            for j in range(K): alpha[t,j] /= scale[t]
            ll += np.log(scale[t]) + peak
        for t in range(n-2,-1,-1):
            for i in range(K):
                v = 0.0
                for j in range(K): v += A[i,j]*b[t+1,j]*beta[t+1,j]
                beta[t,i] = v/scale[t+1]
        for t in range(n):
            z = 0.0
            for k in range(K):
                gamma[lo+t,k] = alpha[t,k]*beta[t,k]
                z += gamma[lo+t,k]
            for k in range(K): gamma[lo+t,k] /= max(z,1e-300)
        for k in range(K): pi_counts[k] += gamma[lo,k]
        for t in range(n-1):
            for i in range(K):
                for j in range(K):
                    transitions[i,j] += alpha[t,i]*A[i,j]*b[t+1,j]*beta[t+1,j]/scale[t+1]
    return gamma, transitions, pi_counts, ll

def log_emissions(counts, returns, q, p, sigma):
    T = len(counts)
    K = len(p)
    out = np.empty((T,K))
    iszero = returns == 0
    for k in range(K):
        out[:,k] = np.log(np.clip(q[k,counts],1e-12,1.))
        out[iszero,k] += np.log(max(p[k],1e-10))
        x = returns[~iszero]
        out[~iszero,k] += (np.log(max(1-p[k],1e-10))
                  - np.log(sigma[k]) - 0.5*np.log(2*np.pi)
                  - 0.5*(x/sigma[k])**2)
    return out

def fit_hmm(counts, returns, starts, ends, K=2, seed=3, max_iter=45, tol=1e-5):
    rng = np.random.default_rng(seed)
    q = rng.dirichlet([1,1,1,1],size=K)
    overall_zero = np.mean(returns==0)
    p = np.clip(overall_zero + rng.normal(0,0.06,K),0.1,0.999)
    nz = returns[returns != 0]
    baseline = max(np.std(nz),0.001)
    sigma = baseline*np.exp(rng.normal(0,0.5,K))
    A = np.full((K,K),0.10/(K-1))
    np.fill_diagonal(A,0.90)
    pi = np.ones(K)/K
    log_likelihood=[]
    for iteration in range(max_iter):
        logb = log_emissions(counts,returns,q,p,sigma)
        gamma, xi, pi_counts, ll = e_step(logb,A,pi,starts,ends)
        log_likelihood.append(ll)
        if iteration > 0 and abs(ll-log_likelihood[-2]) < tol*(1+abs(ll)):
            break
        # M-step
        weights = gamma.sum(axis=0)
        for k in range(K):
            by_cat = np.bincount(counts,weights=gamma[:,k],minlength=4)
            q[k] = (by_cat+1e-3)/(weights[k]+4e-3)
            p[k] = np.clip(gamma[returns==0,k].sum()/weights[k],1e-5,1-1e-5)
            g_nonzero = gamma[returns!=0,k]
            sigma[k] = np.sqrt(max(np.sum(g_nonzero*nz**2) / max(g_nonzero.sum(),1e-12), 1e-8))
        A = (xi+1e-4)/(xi.sum(axis=1,keepdims=True)+K*1e-4)
        pi = (pi_counts+1e-4)/(pi_counts.sum()+K*1e-4)
    # final E-step with final parameters
    gamma, xi, pi_counts, ll = e_step(log_emissions(counts,returns,q,p,sigma),A,pi,starts,ends)
    return dict(q=q,p=p,sigma=sigma,A=A,pi=pi,gamma=gamma,
                ll=ll,history=np.array(log_likelihood),iterations=len(log_likelihood))
```

```python
def as_arrays(frame):
    sizes = frame.groupby(['symbol','date'],sort=False).size().to_numpy()
    ends=np.cumsum(sizes).astype(np.int64)
    starts=np.r_[0,ends[:-1]].astype(np.int64)
    return (frame['count_cat'].to_numpy(np.int64),
            frame['x_bp'].to_numpy(np.float64),starts,ends)
trc,trx,trs,tre=as_arrays(train)
tec,tex,tes,tee=as_arrays(test)

def score_holdout(m, counts, returns, starts, ends):
    # The likelihood computation is forward-only mathematically, even though
    # our e_step also runs backward to return diagnostic arrays.
    b=log_emissions(counts,returns,m['q'],m['p'],m['sigma'])
    return e_step(b,m['A'],m['pi'],starts,ends)[3]

models={}
rows=[]
for K in [2,3,4]:
    print('Fitting K =',K)
    runs=[fit_hmm(trc,trx,trs,tre,K=K,seed=seed,max_iter=55) for seed in [3,11,29]]
    m=max(runs,key=lambda x:x['ll'])
    heldout=score_holdout(m,tec,tex,tes,tee)
    models[K]=m
    # Parameter count: K-1 initial, K(K-1) transitions, 3K categorical count,
    # K zero-mass and K Gaussian SD -> K^2+5K-1
    nparam=(K-1)+K*(K-1)+3*K+K+K
    rows.append({'K':K,'train_ll_per_second':m['ll']/len(train),
                 'holdout_ll_per_second':heldout/len(test),
                 'train_BIC':-2*m['ll']+nparam*np.log(len(train)),
                 'EM_iterations':m['iterations']})
results=pd.DataFrame(rows).sort_values('K')
display(results)
# Prefer future-day likelihood over in-sample fit. Note that selection on this
# holdout reuses it as a validation set, not an untouched final test set.
best_K=int(results.loc[results['holdout_ll_per_second'].idxmax(),'K'])
print('Selected K by held-out likelihood:',best_K)
```

```python
m=models[best_K]
activity=m['q']@np.array([0.,1.,2.,3.]) # 3 denotes capped 3+, not actual mean
order=np.argsort(-activity)
for key in ['q','p','sigma','pi']:
    m[key]=m[key][order]
m['A']=m['A'][np.ix_(order,order)]
m['gamma']=m['gamma'][:,order]
summary=pd.DataFrame({
  'state':np.arange(1,best_K+1),
  'P(count=0)':m['q'][:,0], 'P(count=1)':m['q'][:,1],
  'P(count=2)':m['q'][:,2], 'P(count>=3)':m['q'][:,3],
  'P(zero return)':m['p'], 'SD of nonzero return (bp)':m['sigma'],
  'unconditional one-second SD (bp)':m['sigma']*np.sqrt(1-m['p']),
  'P(stay)':np.diag(m['A']),
  'expected run (sec)':1/(1-np.diag(m['A']))
})
display(summary.round(5))
display(pd.DataFrame(m['A'],index=[f'State {i}' for i in summary.state],
                     columns=[f'State {i}' for i in summary.state]).round(4))
```

```python
def filter_days(counts, returns, starts, ends, model):
    K=len(model['p']); out=np.empty((len(counts),K))
    lb=log_emissions(counts,returns,model['q'],model['p'],model['sigma'])
    for lo,hi in zip(starts,ends):
        prior=model['pi'].copy()
        for t in range(lo,hi):
            if t>lo: prior=out[t-1]@model['A']
            logpost=np.log(np.maximum(prior,1e-300))+lb[t]
            out[t]=np.exp(logpost-logsumexp(logpost))
    return out
prob=filter_days(tec,tex,tes,tee,m)
for k in range(best_K): test[f'prob_state_{k+1}']=prob[:,k]
test['state_filtered']=prob.argmax(axis=1)+1
test['confidence_filtered']=prob.max(axis=1)
print('Held-out regime fractions:')
display(test['state_filtered'].value_counts(normalize=True).sort_index().rename('fraction').to_frame())
print('Uncertain seconds (confidence < 0.7):',round((test['confidence_filtered']<0.7).mean(),4))
example_day=sorted(test_days)[0]
cols=['ts_sec','mid','tick_count','x_bp','state_filtered','confidence_filtered']+[f'prob_state_{k+1}' for k in range(best_K)]
display(test.loc[test['date']==example_day,cols].iloc[::300].head(30).round(4))
```

```python
@njit
def e_step(logb, A, pi, starts, ends):
    # Scaled forward-backward over independent trading-day sequences.
    T,K = logb.shape
    gamma = np.empty((T,K))
    transitions = np.zeros((K,K))
    pi_counts = np.zeros(K)
    ll = 0.0
    for seq in range(len(starts)):
        lo,hi = starts[seq],ends[seq]
        n = hi-lo
        alpha = np.zeros((n,K))
        beta = np.ones((n,K))
        b = np.empty((n,K))
        scale = np.empty(n)
        for t in range(n):
            peak = np.max(logb[lo+t])
            for k in range(K): b[t,k] = np.exp(logb[lo+t,k]-peak)
            for j in range(K):
                if t == 0: alpha[t,j] = pi[j] * b[t,j]
                else:
                    v = 0.0
                    for i in range(K): v += alpha[t-1,i] * A[i,j]
                    alpha[t,j] = b[t,j] * v
            s = np.sum(alpha[t])
            scale[t] = max(s,1e-300)
            for j in range(K): alpha[t,j] /= scale[t]
            ll += np.log(scale[t]) + peak
        for t in range(n-2,-1,-1):
            for i in range(K):
                v = 0.0
                for j in range(K): v += A[i,j]*b[t+1,j]*beta[t+1,j]
                beta[t,i] = v/scale[t+1]
        for t in range(n):
            z = 0.0
            for k in range(K):
                gamma[lo+t,k] = alpha[t,k]*beta[t,k]
                z += gamma[lo+t,k]
            for k in range(K): gamma[lo+t,k] /= max(z,1e-300)
        for k in range(K): pi_counts[k] += gamma[lo,k]
        for t in range(n-1):
            for i in range(K):
                for j in range(K):
                    transitions[i,j] += alpha[t,i]*A[i,j]*b[t+1,j]*beta[t+1,j]/scale[t+1]
    return gamma, transitions, pi_counts, ll
```

```python
def joint_log_emissions(counts, returns, q, p_by_count, sigma):
    N=len(counts); K=len(sigma)
    assert not np.any((counts==0)&(returns!=0)), 'structural zero violated'
    zero=returns==0
    out=np.empty((N,K),dtype=float)
    for k in range(K):
        out[:,k]=np.log(np.maximum(q[k,counts],1e-300))
        out[zero,k]+=np.log(np.maximum(p_by_count[k,counts[zero]],1e-300))
        nz=~zero
        x=returns[nz]
        out[nz,k]+=(np.log(np.maximum(1-p_by_count[k,counts[nz]],1e-300))
                       -np.log(sigma[k])-0.5*np.log(2*np.pi)
                       -0.5*(x/sigma[k])**2)
    return out

def fit_joint(counts,returns,starts,ends,K=4,seed=1,max_iter=65,tol=1e-6):
    rng=np.random.default_rng(seed)
    q=rng.dirichlet(np.ones(4)*2,size=K)
    # Initialize heterogeneous count profiles and zero probabilities
    centers=np.linspace(0,3,K).astype(int)
    for k,c in enumerate(centers): q[k,c]+=2; q[k]/=q[k].sum()
    p=np.empty((K,4))
    p[:,0]=1.0
    base_p=np.array([1.0]+[np.mean(returns[counts==c]==0) if np.any(counts==c) else .5 for c in (1,2,3)])
    p[:,1:]=np.clip(base_p[1:]+rng.normal(0,.09,(K,3)),.01,.999)
    nz=returns[returns!=0]
    sig=np.maximum(np.std(nz),0.03)*np.exp(rng.normal(0,.35,K))
    A=np.full((K,K),.12/(K-1)); np.fill_diagonal(A,.88)
    pi=np.full(K,1./K)
    history=[]
    for iteration in range(max_iter):
        b=joint_log_emissions(counts,returns,q,p,sig)
        gamma,trans,pi_counts,ll=e_step(b,A,pi,starts,ends)
        history.append(ll)
        if len(history)>1 and abs(history[-1]-history[-2])<tol*(1+abs(ll)):
            break
        weights=gamma.sum(axis=0)
        nzmask=returns!=0
        for k in range(K):
            bycat=np.bincount(counts,weights=gamma[:,k],minlength=4)
            q[k]=(bycat+0.01)/(weights[k]+.04)
            for c in (1,2,3):
                sel=counts==c
                denom=bycat[c]
                # weak Beta regularization; avoids exactly 0 or 1
                num=gamma[sel & (returns==0),k].sum()
                p[k,c]=(num+0.1)/(denom+0.2)
            p[k,0]=1.0
            w=gamma[nzmask,k]
            sig[k]=np.sqrt(max((w*(returns[nzmask]**2)).sum()/max(w.sum(),1e-12),1e-8))
        A=(trans+1e-3)/(trans.sum(axis=1,keepdims=True)+K*1e-3)
        pi=(pi_counts+1e-3)/(pi_counts.sum()+K*1e-3)
    _,_,_,final_ll=e_step(joint_log_emissions(counts,returns,q,p,sig),A,pi,starts,ends)
    return dict(q=q,p=p,sigma=sig,A=A,pi=pi,ll=final_ll,iterations=len(history),history=history)

# Small unit tests: structural zero, finite valid likelihoods.
q0=np.array([[.5,.3,.15,.05],[.2,.5,.2,.1]])
p0=np.array([[1,.8,.6,.4],[1,.9,.7,.5]])
s0=np.array([.5,1.])
assert np.isfinite(joint_log_emissions(np.array([0,1,3]),np.array([0.,0.,.2]),q0,p0,s0)).all()
print('Emission tests passed.')
```

```python
def validation_loglik(model,counts,rets,starts,ends):
    b=joint_log_emissions(counts,rets,model['q'],model['p'],model['sigma'])
    return e_step(b,model['A'],model['pi'],starts,ends)[3]

models={}; rows=[]
for K in (2,3,4):
    print('Training joint model with K=',K)
    trials=[fit_joint(trc,trx,trs,tre,K=K,seed=s,max_iter=65) for s in (5,17,41)]
    model=max(trials,key=lambda m:m['ll'])
    models[K]=model
    # Free parameters: K-1 initial + K(K-1) transitions + 3K q + 3K p + K sigma
    param_count=K*K+6*K-1
    rows.append(dict(K=K,train_LL_per_sec=model['ll']/len(train),
                     validation_LL_per_sec=validation_loglik(model,vc,vx,vs,ve)/len(valid),
                     BIC=-2*model['ll']+param_count*np.log(len(train)),
                     iterations=model['iterations']))
comparison=pd.DataFrame(rows)
display(comparison.round(5))
best_K=int(comparison.loc[comparison['validation_LL_per_sec'].idxmax(),'K'])
print('Best K on validation likelihood:',best_K)
print('Note: validation is used for K selection; retain a final untouched test period for subsequent assessment.')
```

```python
def filter_joint(counts,returns,starts,ends,m):
    logb=joint_log_emissions(counts,returns,m['q'],m['p'],m['sigma'])
    K=len(m['sigma']); prob=np.empty((len(counts),K))
    for lo,hi in zip(starts,ends):
        for t in range(lo,hi):
            prior=m['pi'] if t==lo else prob[t-1]@m['A']
            z=np.log(np.maximum(prior,1e-300))+logb[t]
            prob[t]=np.exp(z-logsumexp(z))
    return prob
vprob=filter_joint(vc,vx,vs,ve,m)
valid=valid.copy()
valid['state_filtered']=vprob.argmax(axis=1)+1
valid['confidence_filtered']=vprob.max(axis=1)
for k in range(best_K): valid[f'P_state{k+1}_filtered']=vprob[:,k]
print('Validation state fractions:')
display(valid['state_filtered'].value_counts(normalize=True).sort_index().round(4))
print('Uncertain fraction (<70% posterior confidence):',round((valid['confidence_filtered']<.70).mean(),4))
print('P(filtered state | had_update):')
display(pd.crosstab(valid['had_update'],valid['state_filtered'],normalize='index').round(3))
print('P(filtered state | midprice changed):')
display(pd.crosstab(valid['x_bp']!=0,valid['state_filtered'],normalize='index').round(3))
features=[f for f in ['spread_bps','queue_imbalance_l1','depth_imbalance_l5','ofi_l1_sum_60s_signed_log'] if f in valid]
if features: display(valid.groupby('state_filtered')[features].agg(['mean','median']).round(4))
```

```python
def log_emissions(counts, returns, q, p, sigma):
    T = len(counts)
    K = len(p)
    out = np.empty((T,K))
    iszero = returns == 0
    for k in range(K):
        out[:,k] = np.log(np.clip(q[k,counts],1e-12,1.))
        out[iszero,k] += np.log(max(p[k],1e-10))
        x = returns[~iszero]
        out[~iszero,k] += (np.log(max(1-p[k],1e-10))
                  - np.log(sigma[k]) - 0.5*np.log(2*np.pi)
                  - 0.5*(x/sigma[k])**2)
    return out

def fit_independent(counts, returns, starts, ends, K=2, seed=3, max_iter=45, tol=1e-5):
    rng = np.random.default_rng(seed)
    q = rng.dirichlet([1,1,1,1],size=K)
    overall_zero = np.mean(returns==0)
    p = np.clip(overall_zero + rng.normal(0,0.06,K),0.1,0.999)
    nz = returns[returns != 0]
    baseline = max(np.std(nz),0.001)
    sigma = baseline*np.exp(rng.normal(0,0.5,K))
    A = np.full((K,K),0.10/(K-1))
    np.fill_diagonal(A,0.90)
    pi = np.ones(K)/K
    log_likelihood=[]
    for iteration in range(max_iter):
        logb = log_emissions(counts,returns,q,p,sigma)
        gamma, xi, pi_counts, ll = e_step(logb,A,pi,starts,ends)
        log_likelihood.append(ll)
        if iteration > 0 and abs(ll-log_likelihood[-2]) < tol*(1+abs(ll)):
            break
        # M-step
        weights = gamma.sum(axis=0)
        for k in range(K):
            by_cat = np.bincount(counts,weights=gamma[:,k],minlength=4)
            q[k] = (by_cat+1e-3)/(weights[k]+4e-3)
            p[k] = np.clip(gamma[returns==0,k].sum()/weights[k],1e-5,1-1e-5)
            g_nonzero = gamma[returns!=0,k]
            sigma[k] = np.sqrt(max(np.sum(g_nonzero*nz**2) / max(g_nonzero.sum(),1e-12), 1e-8))
        A = (xi+1e-4)/(xi.sum(axis=1,keepdims=True)+K*1e-4)
        pi = (pi_counts+1e-4)/(pi_counts.sum()+K*1e-4)
    # final E-step with final parameters
    gamma, xi, pi_counts, ll = e_step(log_emissions(counts,returns,q,p,sigma),A,pi,starts,ends)
    return dict(q=q,p=p,sigma=sigma,A=A,pi=pi,gamma=gamma,
                ll=ll,history=np.array(log_likelihood),iterations=len(log_likelihood))
```

```python
baseline_trials=[fit_independent(trc,trx,trs,tre,K=best_K,seed=s,max_iter=65) for s in (5,17,41)]
base=max(baseline_trials,key=lambda m:m['ll'])
base_valid=e_step(log_emissions(vc,vx,base['q'],base['p'],base['sigma']),
                  base['A'],base['pi'],vs,ve)[3]
joint_valid=validation_loglik(m,vc,vx,vs,ve)
comparison_ll=pd.DataFrame([{'model':'Independent count + return','K':best_K,
                            'training LL/sec':base['ll']/len(train),'validation LL/sec':base_valid/len(valid)},
                            {'model':'Joint count → conditional return','K':best_K,
                            'training LL/sec':m['ll']/len(train),'validation LL/sec':joint_valid/len(valid)}])
display(comparison_ll.round(5))
print('Joint minus independent validation LL/sec:',round((joint_valid-base_valid)/len(valid),6))
print('For multiple K values, compare both families at each K; matching K alone does not equal matching complexity.')
```

```python
import numpy as np
import pandas as pd

v = v.sort_values(["symbol", "date", "ts_sec"]).copy()

g = v.groupby(["symbol", "date"], sort=False)

# Consecutive one-second log-mid returns, in bp
previous_mid = g["mid"].shift(1)
previous_ts = g["ts_sec"].shift(1)

v["r_bp"] = 10000 * np.log(v["mid"] / previous_mid)

v.loc[
    (v["ts_sec"] - previous_ts != 1)
    | (v["mid"] <= 0)
    | (previous_mid <= 0),
    "r_bp"
] = np.nan

# Compute future realized variance using the next 60 one-second returns
def add_future_rv(group, horizon=60):
    group = group.copy()

    r2 = group["r_bp"].pow(2)

    future_sum = (
        r2.iloc[::-1]
        .rolling(horizon, min_periods=horizon)
        .sum()
        .iloc[::-1]
        .shift(-1)
    )

    future_ts = group["ts_sec"].shift(-horizon)
    valid = (future_ts - group["ts_sec"]).eq(horizon)

    group["future_rv_60"] = future_sum.where(valid)
    return group

v = (
    v.groupby(["symbol", "date"], group_keys=False)
     .apply(add_future_rv)
)

rv_summary = (
    v.groupby("state_filtered")
     .agg(
         n=("future_rv_60", "count"),
         mean_future_rv=("future_rv_60", "mean"),
         median_future_rv=("future_rv_60", "median")
     )
)

display(rv_summary.round(5))
```

```python
import numpy as np
import pandas as pd

from sklearn.linear_model import Ridge
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.metrics import mean_absolute_error, mean_squared_error

data = v.sort_values(["symbol", "date", "ts_sec"]).copy()

# Only information available through time t
g = data.groupby(["symbol", "date"], sort=False)

data["past_rv_60"] = g["r_bp"].transform(
    lambda x: x.pow(2).rolling(60, min_periods=60).sum()
)

data["past_rv_300"] = g["r_bp"].transform(
    lambda x: x.pow(2).rolling(300, min_periods=300).sum()
)

data["past_change_rate_60"] = g["r_bp"].transform(
    lambda x: x.notna().astype(float).where(x.notna())
        .mul(x.ne(0).astype(float))
        .rolling(60, min_periods=60).sum()
) / 60

data["current_update_count"] = (
    data["tick_count"].clip(lower=0)
)

base_features = [
    "current_update_count",
    "past_rv_60",
    "past_rv_300",
    "past_change_rate_60",
]

# Filtered state: information available at time t
state_dummies = pd.get_dummies(
    data["state_filtered"],
    prefix="state",
    dtype=float
)

data = pd.concat([data, state_dummies], axis=1)

hmm_features = base_features + list(state_dummies.columns)

required = list(set(
    base_features +
    ["future_rv_60", "state_filtered"]
))

data = data.dropna(subset=required).copy()

print("Usable observations:", len(data))
print("Available days:", data["date"].nunique())
```

```python
# Chronological division by date
dates = sorted(data["date"].unique())

split = int(len(dates) * 0.6)

train_dates = dates[:split]
test_dates = dates[split:]

train = data[data["date"].isin(train_dates)]
test = data[data["date"].isin(test_dates)]

assert len(train) > 0 and len(test) > 0

def evaluate_model(features, label):
    model = make_pipeline(
        StandardScaler(),
        Ridge(alpha=10.0)
    )

    # Log transform to reduce target skew
    y_train = np.log1p(train["future_rv_60"])

    model.fit(train[features], y_train)

    pred = np.expm1(model.predict(test[features]))
    pred = np.maximum(pred, 0)

    actual = test["future_rv_60"].to_numpy()

    return {
        "model": label,
        "MAE": mean_absolute_error(actual, pred),
        "RMSE": np.sqrt(mean_squared_error(actual, pred)),
        "Log_MAE": mean_absolute_error(
            np.log1p(actual),
            np.log1p(pred)
        )
    }

results = pd.DataFrame([
    evaluate_model(base_features, "Baseline"),
    evaluate_model(hmm_features, "Baseline + HMM")
])

display(results.round(5))

baseline_mae = results.loc[0, "MAE"]
hmm_mae = results.loc[1, "MAE"]

print(
    "HMM MAE improvement (%):",
    round(100 * (baseline_mae - hmm_mae) / baseline_mae, 2)
)
```
